J’ai d’abord fait une pré-labellisation automatique par règles de mots-clés.

L’idée était simple : pour chaque famille de problème, on définit une liste de mots ou expressions associés.

Par exemple, pour le remboursement :

remboursé
remboursement
rembourser
avoir
bon d’achat

Si un avis négatif contient ces mots, on active :

pb_remboursement_avoir = 1

Pour la livraison :

livraison
colis
retard
transporteur
point relais
livreur

On active :

pb_livraison_colis_retard = 1

Pour le service client :

service client
SAV
réclamation
appel
mail
réponse
relance
contact

On active :

pb_service_client_sav = 1

Donc la méthode repose sur une logique explicable :

Si un avis négatif contient des termes associés à un problème, alors ce problème est marqué comme présent.

Pourquoi on a filtré sur les avis négatifs ?

J’ai appliqué la labellisation principalement sur les avis avec :

rating <= 2

Parce que le but était d’identifier ce qui ne va pas.

C’est important, parce qu’un avis positif peut contenir des mots comme :

livraison rapide
service client efficace
produit conforme

Si on ne filtre pas, le simple mot “livraison” pourrait déclencher un faux problème de livraison, alors que le commentaire est positif.

Donc le filtre rating <= 2 permet de se concentrer sur les avis réellement critiques.

```python
import pandas as pd

df = pd.read_csv("all_reviews_multilabel.csv", encoding="utf-8-sig")

label_cols = [col for col in df.columns if col.startswith("pb_")]

df_neg = df[df["rating"] <= 2].copy()

resume = (
    df_neg
    .groupby("concurrent")[label_cols]
    .mean()
    .mul(100)
    .round(1)
)

resume
```

```python
#28.2 % des avis négatifs de boulanger mentionnent un problème de remboursement.
```

```python
probleme_principal = resume.idxmax(axis=1)
score_principal = resume.max(axis=1)

top_probleme = pd.DataFrame({
    "probleme_principal": probleme_principal,
    "pourcentage": score_principal
})

top_probleme
```

```python
top3_par_concurrent = (
    resume
    .stack()
    .reset_index()
)

top3_par_concurrent.columns = ["concurrent", "probleme", "pourcentage"]

top3_par_concurrent = (
    top3_par_concurrent
    .sort_values(["concurrent", "pourcentage"], ascending=[True, False])
    .groupby("concurrent")
    .head(3)
)

top3_par_concurrent
```

```python
#Donc le SAV devient souvent le label dominant, parce qu’il accompagne presque tous les autres problèmes.

#Chez Boulanger, les avis négatifs mentionnent très fortement le service client et les problèmes de commande, d’annulation ou de stock. Les problèmes de produit défectueux ou non conforme arrivent aussi très haut.

#Le fait que service_client_sav et commande_annulation_stock soient tous les deux à pp 60 % veut dire qu’ils sont aussi fréquents l’un que l’autre dans les avis négatifs Boulanger.



#L’analyse multi-label des avis négatifs montre que le service client / SAV constitue un irritant transversal pour 
# l’ensemble des concurrents étudiés. Ce résultat doit toutefois être interprété avec prudence, 
# car le service client est souvent mentionné après un autre problème initial : commande annulée, livraison en retard, 
# produit défectueux ou remboursement non reçu.

#Des différences apparaissent néanmoins selon les enseignes. 
#Boulanger et Cdiscount sont davantage associés aux problèmes de commande, d’annulation ou de stock, 
#ainsi qu’aux produits défectueux ou non conformes. Conforama et Fnac font ressortir un trio service client, commande et livraison. 
#Darty se distingue davantage par les problèmes de produit et de livraison.
```

```python
#top sans prendre en compte le SAV
label_cols_sans_sav = [
    col for col in label_cols 
    if col != "pb_service_client_sav"
]

resume_sans_sav = (
    df_neg
    .groupby("concurrent")[label_cols_sans_sav]
    .mean()
    .mul(100)
    .round(1)
)

probleme_principal_sans_sav = pd.DataFrame({
    "probleme_principal_sans_sav": resume_sans_sav.idxmax(axis=1),
    "pourcentage": resume_sans_sav.max(axis=1)
})

probleme_principal_sans_sav
```

J’ai d’abord transformé les commentaires clients en données exploitables grâce à une approche de labellisation multi-label. L’objectif était d’identifier les différentes familles de problèmes évoquées dans les avis négatifs, comme le service client, la livraison, le remboursement, les produits défectueux ou les problèmes de commande.

J’ai choisi une approche multi-label parce qu’un même avis peut contenir plusieurs irritants. Par exemple, un client peut se plaindre à la fois d’un retard de livraison, d’un service client injoignable et d’un remboursement non reçu.

La première labellisation a été réalisée de manière automatique à partir de règles de mots-clés. Chaque catégorie de problème a été associée à des termes caractéristiques. Les labels ont ensuite été appliqués uniquement aux avis négatifs, c’est-à-dire aux avis ayant une note inférieure ou égale à 2, afin de limiter les faux positifs.

Cette méthode a été retenue car elle est explicable, contrôlable et adaptée à une première analyse concurrentielle. Elle permet de comparer les enseignes sur les problèmes les plus fréquemment mentionnés, tout en gardant la possibilité de vérifier et corriger les labels manuellement avant un éventuel entraînement de modèle de machine learning.
